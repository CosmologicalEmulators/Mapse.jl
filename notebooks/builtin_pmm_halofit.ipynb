# Builtin MAPSE emulator and Halofit `Pmm`

This notebook shows the current artifact-backed workflow:

1. load the linear total-matter emulator from the official `mnuw0wacdm_class` artifact;
2. compute the linear `Pmm(k,z)`;
3. compute the nonlinear `Pmm` using Halofit.


```julia
import Pkg
Pkg.activate(".")
using Mapse
```

```julia
root = Mapse.artifact_path(Mapse.DEFAULT_EMULATOR_ARTIFACT)
pmm_emu = Mapse.load_emulator(joinpath(root, "Pk_lin_mm"))

(
    artifact = Mapse.DEFAULT_EMULATOR_ARTIFACT,
    k_points = length(Mapse.get_kgrid(pmm_emu)),
    description = Mapse.get_emulator_description(pmm_emu),
)
```

```julia
params = [3.044, 0.9649, 67.36, 0.02237, 0.12, 0.06, -1.0, 0.0]
z = [0.0, 0.5, 1.0]
D = [1.0, 0.772, 0.606]

k_lin = Mapse.get_kgrid(pmm_emu)
Pmm_lin = Mapse.get_Pk(params, z, D, pmm_emu)

(
    Pmm_shape = size(Pmm_lin),
    k_points = length(k_lin),
    k_range = (first(k_lin), last(k_lin)),
)
```

```julia
cpar = Mapse.halofit_cosmology(params)
Ωm_z, Ωv_z = Mapse.halofit_background(cpar, z)
Pmm_nl = Mapse.halofit_pmm(cpar, z, k_lin, Pmm_lin, Ωm_z, Ωv_z)

(
    Pmm_shape = size(Pmm_nl),
    k_points = length(k_lin),
    k_range = (first(k_lin), last(k_lin)),
)
```
